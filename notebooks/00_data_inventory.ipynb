# 00. Data inventory

This notebook reads the machine-generated inventory. It does not infer PE labels, priors, or injection normalization.

```python
from pathlib import Path
import json
import os
import pandas as pd

results_root = Path(os.environ["GWTC_RESULTS_ROOT"])
inventory_dir = results_root / "pilot" / "inventory"
inventory_path = inventory_dir / "data_inventory.json"
inventory_path
```

```python
report = json.loads(inventory_path.read_text())
print("All manifest files present:", report["all_manifest_files_present"])
print("Event-list audit passed:", report["event_list_audit"]["passed"])
pd.DataFrame(report["files"])
```

```python
missing = pd.DataFrame(report["files"]).query("not present")
missing[["role", "filename", "path"]] if len(missing) else "No missing files"
```

```python
for role, item in report["hdf5_reports"].items():
    print("\n", role)
    generic = item.get("generic_hdf5", item.get("inventory", item))
    print("nodes:", generic.get("node_count"), "truncated:", generic.get("truncated"))
```
